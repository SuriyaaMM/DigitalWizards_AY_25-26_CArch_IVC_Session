```python
import jax
import jax.numpy as jnp
import jax.random as jrandom

import time
import os
from IPython.display import display

jax.default_backend()
key = jrandom.key(1)

os.environ["XLA_PYTHON_CLIENT_PREALLOCATE"] = "false"
os.environ["XLA_PYTHON_CLIENT_ALLOCATOR"] = "platform"
```

```python
N = 1_00_00_000
x = jnp.arange(0, N).astype(jnp.float32)
display(x[:10], x[-5:-1])
```

```python
def f(x: jax.Array):
    return jnp.pow(x, 2) - 8

fdash = jax.grad(f)

def newton_method(current_guess: jax.Array, _):
    new_guess = current_guess - f(current_guess) / fdash(current_guess)
    return new_guess, new_guess

NITERATIONS = 100000
```

```python
guess = 1.0

begin = time.perf_counter()
for i in range(NITERATIONS):
    guess = guess - f(guess)/ fdash(guess)
end = time.perf_counter() 

display(f"Elapsed Time = {end - begin}")
display(f"Final root = {guess}")
```

```python
initial_guess = 1.0
begin = time.perf_counter()
final_root, history = jax.lax.scan(newton_method, initial_guess, None, length=100000)
end = time.perf_counter() 

display(f"Elapsed Time = {end - begin}")
display(f"Final root = {final_root}")
display(f"History (size) = {history.size}")
```
