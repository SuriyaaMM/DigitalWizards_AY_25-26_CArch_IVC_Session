```python
import math
import random
import time
from pprint import pprint

# for simplicity say we have only 1,00,00,000 products
prices = [float(random.randint(1, 70_000)) for _ in range(1_00_00_000)]
pprint(prices[:5])

begin = time.perf_counter()
for i in range(len(prices)):
  # apply 20% discount to all elements
  prices[i] = (prices[i] * 0.2)
end = time.perf_counter()

pprint(prices[:5])
print(f"Time taken: {end - begin}")
```
